## Resumo

- Lista = coleção ordenada e **modificável**, entre colchetes. O índice **começa em 0**.
- Métodos: `append`, `insert`, `remove`, `pop`, `sort`; funções: `len`, `sum`, `min`, `max`.
- Percorra listas com `for item in lista`.

### Exemplo 1 – Índices, fatias e tamanho.

Arquivo: `01_listas_basico.py`

```python
destinos = ["Parintins", "Novo Airão", "Presidente Figueiredo", "Maués"]

print(destinos[0])      # primeiro
print(destinos[-1])     # último
print(destinos[1:3])    # fatia
print(len(destinos))
```

### Exemplo 2 – Adicionar, ordenar e percorrer.

Arquivo: `02_metodos_de_listas.py`

```python
peixes = ["Tambaqui", "Pirarucu", "Tucunaré"]
peixes.append("Aruanã")
peixes.sort()
for peixe in peixes:
    print("-", peixe)
```

### Exemplo 3 – Maior, menor e média de uma lista numérica.

Arquivo: `03_media_dos_gastos.py`

```python
gastos = [990.00, 1250.50, 780.00]

for gasto in gastos:
    print(f"R$ {gasto:.2f}")

print(f"Maior: {max(gastos)} | Menor: {min(gastos)}")
print(f"Média: {sum(gastos) / len(gastos):.2f}")
```
