### Exemplo 1 – Cadastro de clientes em memória.

Arquivo: `01_lista_de_dicionarios.py`

**Entradas sugeridas:** ` marina costa ` → `parintins` → `3` → `s` → `joão pedro` → `maués` → `4` → `n`

```python
clientes = []

continuar = "s"
while continuar == "s":
    cliente = {
        "nome": input("Nome: ").strip().title(),
        "destino": input("Destino: ").strip().title(),
        "dias": int(input("Dias: "))
    }
    clientes.append(cliente)
    continuar = input("Cadastrar outro? (s/n): ").lower()

print()
for cl in clientes:
    print(f"{cl['nome']} -> {cl['destino']} ({cl['dias']} dias)")
print(f"Total de clientes: {len(clientes)}")
```
