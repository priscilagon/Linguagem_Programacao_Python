## Resumo

- Métodos: `upper`, `lower`, `title`, `strip`, `split`, `replace`, `find`; função `len`.
- Encadeie métodos: `nome.strip().title()`.
- Validações simples checam apenas o **formato** — use sempre dados fictícios.

### Exemplo 1 – Métodos de texto mais usados.

Arquivo: `01_funcoes_de_texto.py`

```python
nome = "   maria da SILVA  "
print(nome.strip().title())
print("Manaus-AM".split("-"))
print("Novo Airão".replace(" ", "_"))
print("Parintins"[0:3])
print(len("Parintins"))
```

### Exemplo 2 – Validar e-mail e CPF fictício pelo formato.

Arquivo: `02_validacoes_de_texto.py`

**Entradas sugeridas:** `marina@email.com` → `12345678901`

```python
email = input("E-mail: ").strip().lower()
if "@" in email and "." in email:
    print("E-mail com formato aceitável.")
else:
    print("E-mail inválido.")

cpf = input("CPF fictício (só números): ")
if len(cpf) == 11 and cpf.isdigit():
    print("CPF com formato válido.")
else:
    print("CPF deve ter 11 dígitos numéricos.")
```

### Exemplo 3 – Recibo e código de reserva.

Arquivo: `03_recibo_com_f_strings.py`

```python
nome = "marina costa"
destino = "Parintins"
dias = 3
total = 990.00

codigo = f"{destino[:3].upper()}-{dias:02d}-{nome[:2].upper()}"

print("=" * 34)
print("     RECIBO - AMAZÔNIA TUR DIGITAL")
print("=" * 34)
print(f"Cliente: {nome.title()}")
print(f"Destino: {destino}")
print(f"Duração: {dias} dias")
print(f"Total:   R$ {total:.2f}")
print(f"Reserva: {codigo}")
```
