## Resumo

- Modos: `"r"` leitura, `"w"` escrita (**apaga o anterior**), `"a"` acréscimo.
- Use `with open(..., encoding="utf-8") as arquivo:` — o arquivo é fechado automaticamente.
- Para guardar vários dados por linha, use um separador (`;`).
- **Colab:** os arquivos somem ao fim da sessão (monte o Google Drive). **VS Code:** abra a **pasta** do projeto.

### Exemplo 1 – Gravar uma lista e ler de volta.

Arquivo: `01_escrever_e_ler.py`

```python
destinos = ["Parintins", "Novo Airão", "Maués"]

with open("destinos.txt", "w", encoding="utf-8") as arquivo:
    for destino in destinos:
        arquivo.write(destino + "\n")

with open("destinos.txt", "r", encoding="utf-8") as arquivo:
    for linha in arquivo:
        print(linha.strip())
```

### Exemplo 2 – Persistência de registros com separador `;`.

Arquivo: `02_salvar_e_carregar_clientes.py`

```python
def salvar_clientes(clientes, arquivo="clientes.txt"):
    with open(arquivo, "w", encoding="utf-8") as f:
        for cl in clientes:
            f.write(f"{cl['nome']};{cl['destino']};{cl['dias']}\n")

def carregar_clientes(arquivo="clientes.txt"):
    clientes = []
    try:
        with open(arquivo, "r", encoding="utf-8") as f:
            for linha in f:
                nome, destino, dias = linha.strip().split(";")
                clientes.append({"nome": nome, "destino": destino, "dias": int(dias)})
    except FileNotFoundError:
        pass            # primeira execução: o arquivo ainda não existe
    return clientes

clientes = [{"nome": "Marina Costa", "destino": "Parintins", "dias": 3}]
salvar_clientes(clientes)
print(carregar_clientes())
```
