**Habilidade:** utilizar comandos de controle de fluxo

## Resumo

- `break` interrompe o laço; `continue` pula para a próxima repetição; `pass` reserva um bloco vazio.
- `while True` + `break` é a base de qualquer **menu interativo**.
- A **AV2** (duplas) avalia funcionamento, validação de entrada, controle de fluxo e colaboração.

### Exemplo 1 – Pular e interromper repetições.

Arquivo: `01_break_e_continue.py`

```python
for numero in range(1, 11):
    if numero == 4:
        continue        # pula o 4
    if numero == 8:
        break           # encerra no 8
    print(numero)
```

### Exemplo 2 – Menu interativo com `while True`.

Arquivo: `02_menu_while_true.py`

**Entradas sugeridas:** `2` → `9` → `0`

```python
destinos = ["Parintins", "Novo Airão", "Presidente Figueiredo", "Maués"]

while True:
    print("\n=== AGÊNCIA AMAZÔNIA TUR DIGITAL ===")
    print("1 - Calcular viagem")
    print("2 - Ver destinos")
    print("0 - Sair")
    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        pass    # será implementado depois
    elif opcao == "2":
        for destino in destinos:
            print("-", destino)
    elif opcao == "0":
        print("Até logo!")
        break
    else:
        print("Opção inválida!")
        continue
```
