**Objeto do conhecimento:** tipos de dados, declaração de variáveis, `print()` e `input()`

**Habilidade:** utilizar os princípios e funções básicas da linguagem Python

## Resumo

- Variável = nome que guarda um valor. O sinal `=` significa “guarde o valor da direita na variável da esquerda”.
- Tipos básicos: `int`, `float`, `str` e `bool`. Use `type()` para descobrir o tipo.
- `input()` **sempre devolve texto**: converta com `int()` ou `float()` antes de calcular.
- Use **f-strings** para misturar texto e variáveis: `f"{nome} viaja por {dias} dias"`.

### Exemplo 1 – Variáveis e os quatro tipos básicos.

Arquivo: `01_variaveis_e_tipos.py`

```python
destino = "Novo Airão"
dias = 4
diaria = 180.50
confirmado = True

print(destino, dias, diaria, confirmado)
print(type(destino))
print(type(dias))
print(type(diaria))
print(type(confirmado))
```

### Exemplo 2 – Entrada de dados com `input()` e f-string.

Arquivo: `02_entrada_de_dados.py`

**Entradas sugeridas:** `Marina` → `Presidente Figueiredo` → `3`

```python
nome = input("Nome do cliente: ")
destino = input("Destino: ")
dias = int(input("Dias de viagem: "))   # converte para inteiro

print(f"{nome} viajará para {destino} por {dias} dias.")
```

### Exemplo 3 – Conversões com `int()`, `float()` e `str()`.

Arquivo: `03_conversao_de_tipos.py`

```python
texto = "42"
numero = int(texto)
print(numero + 8)

preco = float("180.50")
print(f"Preço: R$ {preco:.2f}")

print("Dias: " + str(3))
```
