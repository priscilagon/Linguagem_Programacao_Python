## Resumo

- Comparações resultam em `True` ou `False`. Atenção: `=` **guarda**, `==` **compara**.
- `and` exige as duas condições; `or` basta uma; `not` inverte.
- Todo programa segue **entrada → processamento → saída**.
- A **Calculadora de viagem v1** é a entrega da Fase 1 (veja o projeto integrador).

### Exemplo 1 – Comparações e operadores lógicos.

Arquivo: `01_relacionais_e_logicos.py`

```python
tem_orcamento = True
tem_vaga = False

print(5 == 5)
print(5 != 3)
print(2 > 4)
print(7 >= 7)
print(tem_orcamento and tem_vaga)
print(tem_orcamento or tem_vaga)
print(not tem_vaga)
print(1 <= 15 <= 30)
```

### Exemplo 2 – Tabela-verdade de `and` e `or`.

Arquivo: `02_tabela_verdade.py`

```python
print("a      b      and    or")
print(True, True, True and True, True or True)
print(True, False, True and False, True or False)
print(False, True, False and True, False or True)
print(False, False, False and False, False or False)
```

### Exemplo 3 – Estrutura de um algoritmo em três blocos.

Arquivo: `03_entrada_processamento_saida.py`

**Entradas sugeridas:** `Parintins` → `3` → `330`

```python
# ----- ENTRADA -----
destino = input("Destino: ")
dias = int(input("Dias: "))
gasto_dia = float(input("Gasto por dia (R$): "))

# ----- PROCESSAMENTO -----
total = dias * gasto_dia

# ----- SAÍDA -----
print(f"Destino: {destino}")
print(f"Total: R$ {total:.2f}")
```
