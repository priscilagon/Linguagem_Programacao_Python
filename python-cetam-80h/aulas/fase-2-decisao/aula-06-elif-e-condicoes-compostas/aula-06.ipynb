## Resumo

- `elif` (senão se) testa várias faixas de cima para baixo e executa **apenas a primeira** verdadeira.
- Combine condições com `and` e `or`.
- Faixas do projeto: Econômico (até R$ 800), Conforto (até R$ 1.500) e Premium (acima).

### Exemplo 1 – Classificação do pacote por faixa de valor.

Arquivo: `01_elif_faixas.py`

**Entradas sugeridas:** `1200`

```python
total = float(input("Total da viagem (R$): "))

if total <= 800:
    pacote = "Econômico"
elif total <= 1500:
    pacote = "Conforto"
else:
    pacote = "Premium"

print(f"Pacote: {pacote}")
```

### Exemplo 2 – Sugestão de destino combinando pacote e dias.

Arquivo: `02_condicoes_compostas.py`

**Entradas sugeridas:** `2` → `300`

```python
dias = int(input("Dias: "))
gasto_dia = float(input("Gasto por dia (R$): "))
total = dias * gasto_dia

if total <= 800:
    pacote = "Econômico"
elif total <= 1500:
    pacote = "Conforto"
else:
    pacote = "Premium"

if pacote == "Econômico" and dias <= 3:
    sugestao = "Presidente Figueiredo"
elif pacote == "Conforto":
    sugestao = "Novo Airão"
else:
    sugestao = "Parintins"

print(f"Total: R$ {total:.2f} - Pacote {pacote}")
print(f"Sugestão de destino: {sugestao}")
```
