## Resumo

- `if` executa um bloco **somente se** a condição for verdadeira; `else` cobre o caso contrário.
- A **indentação** (4 espaços) define o bloco. Não esqueça os dois pontos `:`.
- Valores fixos ficam em MAIÚSCULAS por convenção: `LIMITE = 1000.00`.

### Exemplo 1 – Maior ou menor de idade.

Arquivo: `01_if_else.py`

**Entradas sugeridas:** `20`

```python
idade = int(input("Idade: "))

if idade >= 18:
    print("Maior de idade.")
else:
    print("Menor de idade.")
```

### Exemplo 2 – Condição composta com `or`.

Arquivo: `02_meia_entrada.py`

**Entradas sugeridas:** `65` → `n`

```python
idade = int(input("Idade: "))
estudante = input("É estudante? (s/n): ").lower()
valor = 150.00

if idade >= 60 or estudante == "s":
    valor = valor * 0.5
    print("Meia entrada aplicada.")
else:
    print("Valor integral.")

print(f"Valor do passeio: R$ {valor:.2f}")
```
