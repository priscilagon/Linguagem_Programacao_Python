**Objeto do conhecimento:** operadores `+  -  *  /  //  %  **` e precedência

**Habilidade:** utilizar os princípios e funções básicas da linguagem Python

## Resumo

- `/` sempre devolve decimal; `//` é divisão inteira; `%` é o resto; `**` é potência.
- Precedência: parênteses → `**` → `*  /  //  %` → `+  -`. Na dúvida, use parênteses.
- `:.2f` mostra o número com duas casas decimais — ideal para valores em reais.

### Exemplo 1 – Todos os operadores aritméticos.

Arquivo: `01_operadores.py`

```python
print(7 + 2)     # 9
print(7 - 2)     # 5
print(7 * 2)     # 14
print(7 / 2)     # 3.5
print(7 // 2)    # 3
print(7 % 2)     # 1
print(7 ** 2)    # 49
print(2 + 3 * 4)       # 14
print((2 + 3) * 4)     # 20
```

### Exemplo 2 – Gasto por dia e total da viagem.

Arquivo: `02_custo_da_viagem.py`

**Entradas sugeridas:** `3` → `200` → `50` → `80`

```python
dias = int(input("Quantidade de dias: "))
hospedagem = float(input("Hospedagem por dia (R$): "))
transporte = float(input("Transporte por dia (R$): "))
alimentacao = float(input("Alimentação por dia (R$): "))

gasto_dia = hospedagem + transporte + alimentacao
total = dias * gasto_dia

print(f"Gasto por dia: R$ {gasto_dia:.2f}")
print(f"Total da viagem: R$ {total:.2f}")
```

### Exemplo 3 – Quantos dias cabem no orçamento e quanto sobra.

Arquivo: `03_divisao_inteira_e_resto.py`

```python
orcamento = 1000
diaria = 330
print(f"Dias inteiros: {orcamento // diaria}")
print(f"Sobra: R$ {orcamento % diaria:.2f}")
```
