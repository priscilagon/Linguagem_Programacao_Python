**Habilidade:** aplicar o recurso de modularização e métodos para o uso de funções e parâmetros

## Resumo

- **Parâmetros** estão na definição; **argumentos** são os valores da chamada.
- `print()` **mostra**; `return` **devolve** o valor para o programa.
- Variáveis criadas dentro da função são **locais**. Prefira parâmetros e retorno a variáveis globais.
- Valores padrão: `def validar_orcamento(total, limite=1000.00)`.

### Exemplo 1 – Parâmetros, argumentos e `return`.

Arquivo: `01_parametros_e_retorno.py`

```python
def calcular_total(dias, gasto_dia):
    return dias * gasto_dia

def eh_par(n):
    return n % 2 == 0

total = calcular_total(3, 330.00)
print(total)
print(eh_par(8))
print(eh_par(7))
```

### Exemplo 2 – Valor padrão, busca em dicionário e escopo local.

Arquivo: `02_valor_padrao_e_escopo.py`

```python
def validar_orcamento(total, limite=1000.00):
    return total <= limite

print(validar_orcamento(990))
print(validar_orcamento(1200))
print(validar_orcamento(1200, 1500))

catalogo = {"Parintins": {"preco": 950.00, "dias": 3}, "Maués": {"preco": 1350.00, "dias": 4}}

def buscar_destino(nome, catalogo):
    return catalogo.get(nome.strip().title())

print(buscar_destino("maués", catalogo))
print(buscar_destino("Xingu", catalogo))

def exemplo():
    mensagem = "variável local"
    print(mensagem)

exemplo()
# print(mensagem)   # NameError: a variável não existe fora da função
```
