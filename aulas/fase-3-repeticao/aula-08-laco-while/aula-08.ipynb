## Resumo

- `while` repete **enquanto** a condição for verdadeira — precisa de uma condição de parada.
- Sem atualizar a variável de controle, ocorre **laço infinito** (Ctrl + C no terminal; ■ no Colab).
- Usos: contadores, acumuladores, **validação de entrada** e repetir o programa (`continuar = "s"`).

### Exemplo 1 – Contador com `while`.

Arquivo: `01_while_basico.py`

```python
contador = 1
while contador <= 3:
    print(f"Passageiro {contador}")
    contador = contador + 1
print("Embarque concluído!")
```

### Exemplo 2 – Repete a pergunta até o valor ser válido.

Arquivo: `02_validacao_de_entrada.py`

**Entradas sugeridas:** `40` → `3`

```python
dias = int(input("Dias de viagem (1 a 30): "))
while dias < 1 or dias > 30:
    print("Valor inválido! Digite de 1 a 30.")
    dias = int(input("Dias de viagem (1 a 30): "))
print(f"Dias registrados: {dias}")
```

### Exemplo 3 – Várias viagens em sequência.

Arquivo: `03_repetir_o_programa.py`

**Entradas sugeridas:** `3` → `330` → `s` → `2` → `200` → `n`

```python
continuar = "s"
while continuar == "s":
    dias = int(input("Dias: "))
    gasto_dia = float(input("Gasto por dia: "))
    print(f"Total: R$ {dias * gasto_dia:.2f}")
    continuar = input("Deseja continuar? (s/n): ").lower()
print("Obrigado por usar a calculadora!")
```
