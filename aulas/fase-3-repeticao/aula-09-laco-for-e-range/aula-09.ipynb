**Objeto do conhecimento:** estruturas de repetição: `for`, `range`

**Habilidade:** usar as estruturas/laços de repetição com domínio

## Resumo

- `for` é ideal quando sabemos **quantas vezes** repetir.
- `range(início, fim, passo)`: o **fim não é incluído**.
- Use `for` quando o número de repetições é conhecido e `while` quando depende de uma condição.

### Exemplo 1 – Formas de `range()`.

Arquivo: `01_for_e_range.py`

```python
for i in range(5):
    print(i, end=" ")
print()
for i in range(1, 6):
    print(i, end=" ")
print()
for i in range(0, 10, 2):
    print(i, end=" ")
print()
```

### Exemplo 2 – Contagem regressiva com passo negativo.

Arquivo: `02_contagem_regressiva.py`

```python
for i in range(5, 0, -1):
    print(i)
print("Embarcar!")
```

### Exemplo 3 – Gasto dia a dia com acumulador.

Arquivo: `03_extrato_da_viagem.py`

**Entradas sugeridas:** `3` → `330`

```python
dias = int(input("Dias: "))
gasto_dia = float(input("Gasto por dia (R$): "))
acumulado = 0

for dia in range(1, dias + 1):
    acumulado += gasto_dia
    print(f"Dia {dia}: R$ {gasto_dia:.2f} | Acumulado: R$ {acumulado:.2f}")
```
