## Resumo

- **Classe** é o molde; **objeto** é cada exemplar criado a partir dele.
- **Atributos** guardam dados; **métodos** são as ações. `__init__` prepara o objeto; `self` é o próprio objeto.
- A classe `Usuario` (nome, senha) com `cadastrar()` e `autenticar()` é a base do cofre digital.

### Exemplo 1 – Primeira classe.

Arquivo: `01_classe_cliente.py`

```python
class Cliente:
    def __init__(self, nome, destino):
        self.nome = nome
        self.destino = destino

    def resumo(self):
        return f"{self.nome} viajará para {self.destino}"

cliente1 = Cliente("Marina", "Parintins")
cliente2 = Cliente("João", "Maués")
print(cliente1.resumo())
print(cliente2.resumo())
```

### Exemplo 2 – Classe com três atributos e um método.

Arquivo: `02_classe_pacote.py`

```python
class Pacote:
    def __init__(self, destino, preco, dias):
        self.destino = destino
        self.preco = preco
        self.dias = dias

    def descricao(self):
        return f"{self.destino}: R$ {self.preco:.2f} em {self.dias} dias"

pacote = Pacote("Novo Airão", 1200.00, 4)
print(pacote.descricao())
```

### Exemplo 3 – Classe `Usuario` com persistência em arquivo.

Arquivo: `03_classe_usuario.py`

```python
import os

class Usuario:
    def __init__(self, nome, senha):
        self.nome = nome
        self.senha = senha

    def cadastrar(self, arquivo="usuarios.txt"):
        with open(arquivo, "a", encoding="utf-8") as f:
            f.write(f"{self.nome};{self.senha}\n")

    def autenticar(self, arquivo="usuarios.txt"):
        if not os.path.exists(arquivo):
            return False
        with open(arquivo, "r", encoding="utf-8") as f:
            for linha in f:
                nome, senha = linha.strip().split(";")
                if nome == self.nome and senha == self.senha:
                    return True
        return False

Usuario("marina", "1234").cadastrar()
print(Usuario("marina", "1234").autenticar())
print(Usuario("marina", "errada").autenticar())
```
