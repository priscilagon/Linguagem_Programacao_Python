## Resumo

- **Tupla** `( )`: imutável, para dados fixos.
- **Dicionário** `{chave: valor}`: acesso pela chave, como uma ficha cadastral.
- Percorra com `for chave, valor in dicionario.items()`.
- Dicionários podem conter outros dicionários (catálogo de pacotes).

### Exemplo 1 – Tupla: dados que não mudam.

Arquivo: `01_tuplas.py`

```python
coordenadas_manaus = (-3.119, -60.021)
print(coordenadas_manaus[0])
print(coordenadas_manaus[1])
```

### Exemplo 2 – Criar, alterar, incluir e percorrer.

Arquivo: `02_dicionarios.py`

```python
destino = {"nome": "Parintins", "preco": 950.00, "dias": 3}

print(destino["nome"])
destino["preco"] = 990.00          # altera
destino["transporte"] = "barco"    # inclui

for chave, valor in destino.items():
    print(f"{chave}: {valor}")
```

### Exemplo 3 – Dicionário de dicionários (valores fictícios).

Arquivo: `03_catalogo_de_pacotes.py`

```python
catalogo = {
    "Parintins": {"preco": 950.00, "dias": 3},
    "Novo Airão": {"preco": 1200.00, "dias": 4},
    "Presidente Figueiredo": {"preco": 780.00, "dias": 2},
    "Maués": {"preco": 1350.00, "dias": 4},
}

for nome, dados in catalogo.items():
    print(f"{nome}: R$ {dados['preco']:.2f} em {dados['dias']} dias")
```
